```python
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
from crewai_tools import SerperDevTool, ScrapeWebsiteTool

# 도구 인스턴스 생성
search_tool = SerperDevTool()
scrape_tool = ScrapeWebsiteTool(website_url='https://agnesi.com/ko/recipes/')
```

```python
result = scrape_tool.run()
print(result)
```

```python
from crewai import Agent, Task, Crew, Process

# 요리 추천 Agent 정의
recipe_agent = Agent(
    role="요리 레시피 전문가",
    goal="검색하여 레시피 정리 및 요약",
    backstory="글로벌 요리 전문가 AI 셰프",
    tools=[search_tool, scrape_tool],
    llm='gpt-5.4-mini',
    verbose=True
)

# 태스크 정의
recipe_task = Task(
    description="사이트에서 {food}와 관련된 적합한 레시피를 찾아서 레시피 정보 최대 {count}건 반환해.",
    expected_output="한국어로 작성된 레시피 최대 {count}건의 {food} 레시피 정보 및 링크",
    agent=recipe_agent
)

# Crew 생성 및 실행 (순차적 프로세스 지정)
crew = Crew(
    agents=[recipe_agent],
    tasks=[recipe_task],
    process=Process.sequential,
    verbose=True
)
```

```python
result = await crew.kickoff_async(inputs={'food':'리조또', 'count':3})
print("[요리 레시피 시나리오 응답]\n", result)
```
