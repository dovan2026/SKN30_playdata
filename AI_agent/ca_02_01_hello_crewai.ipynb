```python
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
from crewai import Agent, Task, Crew, Process
from crewai import LLM
```

```python
# LLM 설정
llm = LLM(
    model = 'openai/gpt-4.1-mini',
    temperature = 0.4,
    max_tokens = 1000,              # 생성할 최대 토큰 수
    frequency_penalty = 0.1,        # 단어 반복 억제(높을 수록 중복 감소)
    presence_penalty = -0.1,        # 새로운 주제 언급 유도(높을 수록 다양성 증가) -> 새로운 이야기로 확장될지 여부
    top_p = 0.9                     # 확률 샘플링
)
```

### 1. Agent 생성: role, goal, backstory 설정

```python
agent = Agent(
    role = 'AI 어시스턴트',                     # 에이전트 역할
    goal = '사용자에게 간단한 환영 인사를 제공',   # 에이전트가 달성할 목표
    backstory = '당신은 친절한 AI 비서로, 언제나 정중하고 도움이 되는 인사를 건넵니다.',    # 에이전트 성격/배경
    llm=llm
)
```

### 2. Task 생성: description, expected_output, agent 지정

```python
task = Task(
    description = '사용자에게 환영 인사 한마디를 작성하세요.',      # 에이전트가 수행할 작업 내용
    expected_output = '한 줄의 따뜻한 환영 메시지',               # 기대되는 출력 결과물
    agent = agent                                              # 이 task를 수행할 에이전트
)
```

### 3. Crew 생성: 에이전트와 태스크를 크루로 구성 (순차 프로세스 설정)

```python
# single 에이전트이므로 한개의 팀으로 구성된 crew를 만든 것임.
crew = Crew(
    agent=[agent],                  # 크루에 속한 에이전트 목록
    tasks=[task],                   # 크루가 수행할 태스크 목록
    process=Process.sequential,     # 순차적 지행 방식으로 태스크
    verbose=True                    # 로그 출력
)
```

### 4. Crew 실행: kickoff() 메서드로 태스크 수행 시작

```python
result = await crew.kickoff_async()
```
