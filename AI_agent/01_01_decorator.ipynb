# @decorator

- 파이썬 데코레이터(decorator)는 함수나 메서드를 감싸(wrap)서 그 동작을 확장하거나 수정할 수 있는 문법
- 주로 로깅, 권한 검사, 실행 시간 측정 등 공통 관심사를 함수 본문에 직접 손대지 않고 재사용할 때 유용함
- @데코레이터명 형태로 사용하며 내부적으로는 데코레이터 함수(또는 클래스)에 원본 함수를 인자로 넘겨 새로운 함수를 반환하는 방식으로 동작
- 중복 코드를 줄이고, 전/후 처리를 깔끔히 추가하기 위한 목적
- 예시 : 로그 기록, 권한 확인, 실행 시간 측정 등

- 함수는 객체다 ⇒ 변수에 저장 가능
- 함수를 다른 함수에 인자로 전달하거나, 함수 안에서 새 함수를 반환할 수 있다

```python
def greet(name):
    return f'안녕하세요, {name}'

say = greet
print(say('여러분'))

def call_func(func, arg):
    return func(arg)

print(call_func(greet, '길동이'))
```

```python
def my_decorator(func):             # 이 함수는 다른 함수를 받아서 새로운 함수로 변환
    def wrapper():
        print('함수 실행 전')
        func()
        print('함수 실행 후')
    return wrapper

@my_decorator
def say_hello():                    # say_hello = my_decorator(say_hello)
    print('안녕하세요.')

say_hello()
```

### 연습:
- 두 숫자를 더하는 함수를 다른 함수에 인자로 넘겨 호출해 보기
- 간단한 문자열 처리 함수를 반환하는 함수 작성

### 1. 인자를 받는 함수용 데코레이터

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("함수 실행 전입니다.")
        result = func(*args, **kwargs)
        print("함수 실행 후입니다.")
        return result
    return wrapper

@my_decorator
def greet(name, greeting='안녕하세요'):
    print(f'{name}님, {greeting}')

greet('영희')
greet('철수', greeting='반가워요.')
```

```python
# 1) 두 숫자를 더하는 함수 정의
def add(a, b):
    return a + b

# 2) 어떤 함수를 받아서 숫자를 전달하고 결과를 반환하는 고차 함수 정의
def operate(func, x, y):
    result = func(x, y)
    print(f'결과:{result}')

# 3) add 함수를 operate에 넘겨 실행
operate(add, 10, 20)

```

```python
# 로그용 데코레이터 정의
def operate_decorator(func):
    def wrapper(*args, **kwargs):
        # 호출 전
        print(f"함수 {func.__name__}를 호출합니다: args={args}, kwargs={kwargs}")
        # 실제 함수 실행
        result = func(*args, **kwargs)
        # 호출 후
        print(f"결과: {result}")
        return result
    return wrapper

# 데코레이터 적용
@operate_decorator
def add(a, b):
    """두 숫자를 더해서 반환"""
    return a + b

# 실행
add(10, 20)
# 출력 예시:
# 함수 add를 호출합니다: args=(10, 20), kwargs={}
# 결과: 30
```

### 2.문자열 대문자로 변환하기

```python
def uppercase_decorator(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        if isinstance(result, str):
            return result.upper()
        return result
    return wrapper

@uppercase_decorator
def greet(name):
    return f"Hello, {name}!"

print(greet("world"))  # → "HELLO, WORLD!"
```

### 3.앞뒤 공백 제거하고 출력하기

```python
def strip_args_decorator(func):
    def wrapper(*args, **kwargs):
        # args가 문자열이라면 strip() 적용
        new_args = [arg.strip() if isinstance(arg, str) else arg for arg in args]
        # kwargs도 문자열이면 strip()
        new_kwargs = {k: v.strip() if isinstance(v, str) else v for k, v in kwargs.items()}
        return func(*new_args, **new_kwargs)
    return wrapper

@strip_args_decorator
def echo(text):
    print(f"[Echo] '{text}'")

echo("   안녕, 파이썬!   ")  # → [Echo] '안녕, 파이썬!'
```
