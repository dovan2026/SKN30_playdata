```python
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
from crewai_tools import SerperDevTool, ScrapeWebsiteTool 

# 도구 인스턴스 생성
search_tool = SerperDevTool()       # 공식 문서, ctrl + 클릭
scrape_tool = ScrapeWebsiteTool(website_url='https://www.aitimes.com/news/articleList.html?view_type=sm')
```

```python
result = scrape_tool.run()
result
```

```python
from crewai import Agent, Task, Crew, Process

news_agent = Agent(
    role="{topic} 뉴스 분석가",
    goal="최신 {topic} 뉴스를 검색하여 요약",
    backstory="{topic} 뉴스 분석 전문 AI 기자",
    tools=[search_tool, scrape_tool],
    llm="gpt-5.4-mini",
    verbose=True
)

news_task = Task(
    # description="최신 {topic} 관련 뉴스 {count}건을 요약하고 링크 제공",    # 검색하라는 건지, 스크래핑을 하라는 건지 어쩌라고
    description="최근 {topic} 관련 뉴스와 검색 결과의 뉴스 {count}건을 요약하고 링크 제공",
    expected_output="한국어로 작성된 뉴스 {count}건의 요약 및 링크",
    agent=news_agent
)

crew = Crew(
    agents=[news_agent],
    tasks=[news_task],
    process=Process.sequential,
    verbose=True
)
```

```python
result = await crew.kickoff_async(inputs={'topic': '에이전트', 'count':3})
```
