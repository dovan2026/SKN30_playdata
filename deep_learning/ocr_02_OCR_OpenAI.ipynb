```python
from openai import OpenAI
import base64
import cv2
```

```python
import os
from google.colab import userdata

# Colab Secrets에 저장한 키 불러오기
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
```

```python
# OpenAI API 키 설정
client = OpenAI()
```

```python
# 이미지 파일을 base64로 변환
# 그림을 긴 문자열로 바꿔서 다른 곳에 안전하게 보내기 쉽게 만드는 것.
def encode_image(image_path: str) -> str:
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")
```

```python
# OpenAI API를 사용하여 OCR 수행
def openai_ocr(image_path: str) -> str:
    b64 = encode_image(image_path)

    response = client.responses.create(
        model="gpt-4.1-mini",
        input=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "input_text",
                        "text": "이 이미지에서 보이는 텍스트를 줄바꿈 유지해서 그대로 추출해줘. 번역하지 말고 원문만"
                    },
                    {
                        "type": "input_image",
                        "image_url": f"data:image/jpeg;base64,{b64}"
                    }
                ]
            }
        ],
        temperature=0,
        max_output_tokens=300
    )

    return response.output_text
```

```python
image_path = 'en_sample.jpg'            # OCR을 수행할 이미지 파일
extracted_text = openai_ocr(image_path)
```

```python
# 결과 추출
print('추출된 텍스트:\n', extracted_text)
```
