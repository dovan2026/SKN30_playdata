# CSS 기본 문법 실습

- 목표: HTML 구조에 스타일과 레이아웃을 적용하는 핵심 CSS 문법을 충분히 실습한다.
- 실행 방식: 각 코드 셀은 `%%html`로 HTML과 CSS를 함께 렌더링한다.
- 모든 예제는 독립 실행을 고려해 클래스 이름에 접두사를 사용한다.

# 1. CSS 기본 문법

CSS는 선택자로 대상을 고르고 선언 블록 안에 `속성: 값;`을 작성한다.

```css
선택자 {
  속성: 값;
}
```

````python
%%html
<style>
  .css01-box {
    color: white;
    background-color: #2563eb;
    padding: 16px;
    border-radius: 25px;
  }
</style>

<p class="css01-box">선택자, 속성, 값을 사용해 스타일을 적용합니다.</p>
````

## 1.1 CSS 적용 방법

- 인라인 스타일: 태그의 `style` 속성에 직접 작성
- 내부 스타일: HTML 문서의 `style` 태그에 작성
- 외부 스타일: `.css` 파일을 `link` 태그로 연결

실무에서는 외부 스타일시트를 주로 사용한다.

````python
%%html
<p style="color:#dc2626; font-weight:bold;">인라인 스타일 예시</p>

<style>
  .css01-internal {
    color: #0f766e;
    border-bottom: 2px solid #0f766e;
  }
</style>
<p class="css01-internal">내부 스타일 예시</p>

<pre><code>&lt;link rel="stylesheet" href="style.css"&gt;</code></pre>
````

# 2. 선택자

- 요소 선택자: `p`
- 클래스 선택자: `.card`
- 아이디 선택자: `#main`
- 그룹 선택자: `h1, p`
- 자손 선택자: `.box span`
- 자식 선택자: `.box > span`
- 속성 선택자: `input[type="email"]`

````python
%%html
<style>
  .css02-area p { color: #374151; }
  .css02-area .highlight { background: #fef3c7; }
  .css02-area #important-css-item { color: #dc2626; font-weight: bold; }
  .css02-area ul > li { padding: 4px; }
  .css02-area input[type="email"] { border: 2px solid #2563eb; }
</style>

<div class="css02-area">
  <p>자손 선택자로 색상을 적용했습니다.</p>
  <p class="highlight">클래스 선택자입니다.</p>
  <p id="important-css-item">아이디 선택자입니다.</p>
  <ul><li>자식 선택자 적용</li><li>두 번째 항목</li></ul>
  <input type="email" placeholder="email input">
</div>
````

## 2.1 가상 클래스와 가상 요소

- `:hover`: 마우스를 올렸을 때
- `:focus`: 입력 요소가 선택되었을 때
- `:first-child`, `:last-child`, `:nth-child(n)`: 위치 기반 선택
- `::before`, `::after`: 요소 앞뒤에 가상 콘텐츠 추가

````python
%%html
<style>
  .css02-pseudo li {
    padding: 8px;
    border-bottom: 1px solid #e5e7eb;
  }
  .css02-pseudo li:first-child { font-weight: bold; }
  .css02-pseudo li:nth-child(2) { background: #dbeafe; }
  .css02-pseudo li:last-child { color: #dc2626; }
  .css02-pseudo li::before { content: "✓ "; color: #16a34a; }
  .css02-hover {
    display: inline-block;
    padding: 8px 12px;
    background: #111827;
    color: white;
  }
  .css02-hover:hover { background: #0f766e; }
</style>

<ul class="css02-pseudo">
  <li>첫 번째</li>
  <li>두 번째</li>
  <li>세 번째</li>
</ul>
<button class="css02-hover" type="button">마우스를 올려보세요</button>
````

# 3. 캐스케이딩, 상속, 우선순위

- 나중에 선언한 스타일이 앞선 스타일을 덮을 수 있다.
- 일부 속성은 부모에서 자식으로 상속된다. 예: `color`, `font-family`
- 우선순위는 대략 `인라인 > id > class > element` 순서다.
- `!important`는 가능하면 남용하지 않는다.

````python
%%html
<style>
  .css03-box { color: #2563eb; }
  .css03-box p { color: #16a34a; }
  .css03-box .strong-text { color: #dc2626; }
</style>

<div class="css03-box">
  부모의 color는 자식에게 상속될 수 있습니다.
  <p>요소 선택자가 부모 상속을 덮었습니다.</p>
  <p class="strong-text">클래스 선택자가 더 구체적입니다.</p>
</div>
````

# 4. 색상과 단위

````python
%%html
<style>
  .css04-unit {
    width: min(90%, 520px);
    padding: 1.25rem;
    background: hsl(190, 90%, 92%);
    border-left: 8px solid rgba(37, 99, 235, 0.8);
  }
  .css04-unit strong {
    font-size: 1.5rem;
    color: #1d4ed8;
  }
</style>

<div class="css04-unit">
  <strong>상대 단위 실습</strong>
  <p>width는 화면의 90%를 사용하되 최대 520px까지만 커집니다.</p>
</div>
````

````python
%%html
<style>
  .css10-grid {
    display: grid;
    grid-template-columns: repeat(3, 1fr);
    gap: 10px;
  }
  .css10-grid div {
    padding: 16px;
    background: #e0f2fe;
    border: 1px solid #38bdf8;
    text-align: center;
  }
  .css10-grid .wide { grid-column: span 2; background: #dcfce7; }
</style>

<div class="css10-grid">
  <div class="wide">2칸</div>
  <div>1칸</div>
  <div>1</div>
  <div>2</div>
  <div>3</div>
</div>
````

# 12. Transition과 Transform

- `transition`: 속성 변화에 애니메이션 적용
- `transform`: 이동, 회전, 확대/축소, 기울임
- 버튼, 카드 hover 효과에 자주 사용한다.

````python
%%html
<style>
  .css12-button {
    padding: 12px 18px;
    border: 0;
    border-radius: 8px;
    background: #111827;
    color: white;
    transition: transform .2s, background-color .2s;
  }
  .css12-button:hover {
    background: #0f766e;
    transform: translateY(-3px) scale(1.04);
  }
</style>

<button class="css12-button" type="button">hover 효과</button>
````

# 15. 반응형 웹

- viewport meta 태그가 필요하다.
- `%`, `rem`, `vw`, `vh`, `fr` 같은 상대 단위를 활용한다.
- `@media`로 화면 크기에 따라 스타일을 변경한다.

```html
<meta name="viewport" content="width=device-width, initial-scale=1.0">
```

````python
%%html
<style>
  .css15-responsive {
    display: grid;
    grid-template-columns: repeat(3, 1fr);
    gap: 12px;
  }
  .css15-responsive div {
    padding: 16px;
    background: #fef3c7;
    border: 1px solid #f59e0b;
  }
  @media (max-width: 700px) {
    .css15-responsive {
      grid-template-columns: 1fr;
    }
    .css15-responsive div {
      background: #dbeafe;
      border-color: #2563eb;
    }
  }
</style>

<div class="css15-responsive">
  <div>화면이 넓으면 3열</div>
  <div>좁으면 1열</div>
  <div>색상도 변경</div>
</div>
````

# 17. CSS 수정 과제

1. 카드 레이아웃의 색상 변수를 바꿔본다.
2. `grid-template-columns`의 `minmax` 값을 바꿔 카드 폭 변화를 확인한다.
3. 버튼 hover 효과에 `rotate`, `scale`을 추가한다.
4. 폼 스타일에서 focus 상태를 추가한다.
5. 700px 미디어 쿼리 기준을 500px, 900px로 바꿔 비교한다.
