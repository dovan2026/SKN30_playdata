```python
%%html
<pre id="js04-string"></pre>
<script>
  const rawText = '  Frontend JavaScript  ';
  const lines = [
    `원본: "${rawText}"`,
    `trim: "${rawText.trim()}"`,
    `대문자: ${rawText.trim().toUpperCase()}`,
    `JS 포함 여부: ${rawText.includes('Java')}`,
    `치환: ${rawText.replace('JavaScript', 'JS')}`
  ];
  document.getElementById('js04-string').textContent = lines.join('\n');
</script>
```

# 6. 조건문

```python
%%html
<pre id="js06-if"></pre>
<script>
  const score = 86;
  let grade = '';

  if (score >= 90) {
    grade = 'A';
  } else if (score >= 80) {
    grade = 'B';
  } else if (score >= 70) {
    grade = 'C';
  } else {
    grade = 'D';
  }

  document.getElementById('js06-if').textContent = `점수 ${score}점은 ${grade}등급입니다.`;
</script>
```

## 6.1 switch

- 하나의 값이 여러 경우 중 어디에 해당하는지 비교할 때 사용한다.
- `break`를 적지 않으면 다음 case까지 이어서 실행될 수 있다.

```python
%%html
<pre id="js06-switch"></pre>
<script>
  const day = '화';
  let topic = '';

  switch (day) {
    case '월':
      topic = 'HTML';
      break;
    case '화':
      topic = 'CSS';
      break;
    case '수':
      topic = 'JavaScript';
      break;
    default:
      topic = '복습';
  }

  document.getElementById('js06-switch').textContent = `${day}요일 주제: ${topic}`;
</script>
```

# 7. 반복문

- `for`: 정해진 횟수 반복
- `while`: 조건이 참인 동안 반복
- `for...of`: 배열 값을 하나씩 순회
- `break`: 반복 종료
- `continue`: 이번 반복 건너뛰기

```python
%%html
<pre id="js07-loop"></pre>
<script>
  const lines = [];

  for (let i = 1; i <= 3; i++) {
    lines.push(`for ${i}`);
  }

  let n = 1;
  while (n <= 2) {
    lines.push(`while ${n}`);
    n++;
  }

  for (const topic of ['HTML', 'CSS', 'JS']) {
    lines.push(`topic: ${topic}`);
  }

  document.getElementById('js07-loop').textContent = lines.join('\n');
</script>
```

# 8. 배열

- 배열은 순서가 있는 데이터 묶음이다.
- 인덱스는 0부터 시작한다.
- 주요 메서드: `push`, `pop`, `map`, `filter`, `reduce`, `join`, `includes`

```python
%%html
<pre id="js08-array"></pre>
<script>
  const scores = [75, 88, 92, 64, 100];
  const passed = scores.filter(score => score >= 80);
  const doubled = scores.map(score => score * 2);
  const total = scores.reduce((sum, score) => sum + score, 0);
  const average = total / scores.length;

  const lines = [
    `원본: ${scores.join(', ')}`,
    `80점 이상: ${passed.join(', ')}`,
    `두 배: ${doubled.join(', ')}`,
    `평균: ${average.toFixed(1)}`
  ];
  document.getElementById('js08-array').textContent = lines.join('\n');
</script>
```

# 12. 이벤트

- 이벤트는 클릭, 입력, 변경, 제출 같은 사용자 행동이다.
- `addEventListener(이벤트명, 함수)` 형태로 등록한다.
- 이벤트 객체에는 발생한 이벤트 정보가 들어 있다.

```python
%%html
<button id="js12-click" type="button">클릭</button>
<input id="js12-input" type="text" placeholder="입력해보세요">
<p id="js12-result">이벤트 대기 중</p>

<script>
  const clickButton = document.getElementById('js12-click');
  const input = document.getElementById('js12-input');
  const result = document.getElementById('js12-result');
  let clickCount = 0;

  clickButton.addEventListener('click', function () {
    clickCount++;
    result.textContent = `클릭 ${clickCount}회`;
  });

  input.addEventListener('input', function (event) {
    result.textContent = `입력 값: ${event.target.value}`;
  });
</script>
```

# 13. 폼 이벤트와 기본 동작 막기

- `submit` 이벤트는 폼 제출 시 발생한다.
- `event.preventDefault()`로 기본 제출 동작을 막고 직접 처리할 수 있다.
- 입력값은 `.value`로 읽는다.

```python
%%html
<form id="js13-form">
  <input id="js13-name" type="text" placeholder="이름">
  <button type="submit">인사하기</button>
</form>
<p id="js13-message"></p>

<script>
  const form = document.getElementById('js13-form');
  const nameInput = document.getElementById('js13-name');
  const formMessage = document.getElementById('js13-message');

  form.addEventListener('submit', function (event) {
    event.preventDefault();
    const name = nameInput.value.trim();
    if (name === '') {
      formMessage.textContent = '이름을 입력하세요.';
      formMessage.style.color = '#dc2626';
      return;
    }
    formMessage.textContent = `${name}님, 반갑습니다.`;
    formMessage.style.color = '#0f766e';
  });
</script>
```

# 14. 요소 생성, 추가, 삭제

- `document.createElement()`: 요소 생성
- `append()`, `prepend()`: 요소 추가
- `remove()`: 요소 삭제
- 리스트, 카드, 댓글 UI를 만들 때 자주 사용한다.

```python
%%html
<button id="js14-add" type="button">항목 추가</button>
<button id="js14-remove" type="button">마지막 삭제</button>
<ul id="js14-list"></ul>

<script>
  const addItemButton = document.getElementById('js14-add');
  const removeItemButton = document.getElementById('js14-remove');
  const list = document.getElementById('js14-list');
  let itemNo = 1;

  addItemButton.addEventListener('click', function () {
    const li = document.createElement('li');
    li.textContent = `새 항목 ${itemNo}`;
    list.append(li);
    itemNo++;
  });

  removeItemButton.addEventListener('click', function () {
    const lastItem = list.querySelector('li:last-child');
    if (lastItem) {
      lastItem.remove();
    }
  });
</script>
```

# 15. 타이머

- `setTimeout`: 일정 시간 뒤 한 번 실행
- `setInterval`: 일정 시간마다 반복 실행
- `clearInterval`: 반복 중지

```python
%%html
<button id="js15-start" type="button">타이머 시작</button>
<button id="js15-stop" type="button">중지</button>
<p id="js15-time">0</p>

<script>
  const startTimer = document.getElementById('js15-start');
  const stopTimer = document.getElementById('js15-stop');
  const timeText = document.getElementById('js15-time');
  let timerId = null;
  let seconds = 0;

  startTimer.addEventListener('click', function () {
    if (timerId !== null) return;
    timerId = setInterval(function () {
      seconds++;
      timeText.textContent = `${seconds}초`;
    }, 1000);
  });

  stopTimer.addEventListener('click', function () {
    clearInterval(timerId);
    timerId = null;
  });
</script>
```

# 17. 통합 미니 앱: 할 일 목록

- 입력값 읽기
- submit 이벤트 처리
- 조건문으로 빈 값 검사
- DOM 요소 생성
- 버튼 이벤트로 항목 삭제
- 배열로 데이터 관리

```python
%%html
<style>
  .js17-app {
    max-width: 620px;
    padding: 18px;
    border: 1px solid #d1d5db;
    border-radius: 8px;
    font-family: Arial, sans-serif;
  }
  .js17-app form {
    display: grid;
    grid-template-columns: 1fr auto;
    gap: 8px;
  }
  .js17-app input {
    padding: 10px;
    border: 1px solid #9ca3af;
    border-radius: 6px;
  }
  .js17-app button {
    padding: 10px 12px;
    border: 0;
    border-radius: 6px;
    background: #0f766e;
    color: white;
    cursor: pointer;
  }
  .js17-app ul {
    padding: 0;
    list-style: none;
  }
  .js17-app li {
    display: flex;
    justify-content: space-between;
    align-items: center;
    gap: 10px;
    margin-top: 8px;
    padding: 10px;
    background: #ecfeff;
    border: 1px solid #67e8f9;
    border-radius: 6px;
  }
</style>

<section class="js17-app">
  <h3>할 일 목록</h3>
  <form id="js17-form">
    <input id="js17-input" type="text" placeholder="할 일을 입력하세요">
    <button type="submit">추가</button>
  </form>
  <p id="js17-status">0개 항목</p>
  <ul id="js17-list"></ul>
</section>

<script>
  const todoForm = document.getElementById('js17-form');
  const todoInput = document.getElementById('js17-input');
  const todoList = document.getElementById('js17-list');
  const todoStatus = document.getElementById('js17-status');
  const todos = [];

  function updateStatus() {
    todoStatus.textContent = `${todos.length}개 항목`;
  }

  function renderTodo(text) {
    const li = document.createElement('li');
    const span = document.createElement('span');
    const button = document.createElement('button');

    span.textContent = text;
    button.type = 'button';
    button.textContent = '삭제';
    button.addEventListener('click', function () {
      const index = todos.indexOf(text);
      if (index > -1) {
        todos.splice(index, 1);
      }
      li.remove();
      updateStatus();
    });

    li.append(span, button);
    todoList.append(li);
  }

  todoForm.addEventListener('submit', function (event) {
    event.preventDefault();
    const text = todoInput.value.trim();
    if (text === '') {
      todoInput.style.borderColor = '#dc2626';
      return;
    }
    todoInput.style.borderColor = '#9ca3af';
    todos.push(text);
    renderTodo(text);
    updateStatus();
    todoInput.value = '';
  });
</script>
```
