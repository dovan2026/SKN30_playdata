## 2) text 번역하기
- translator은 translate **패키지**의 핵심 클래스이다. 번역하려는 언어를 설정하고 번역을 수행한다.

    - to_lang: 번역할 목표 언어 설정
    - from_lang: 원본 텍스트의 언어 설정
    - translate(): 번역하려는 텍스트를 입력받아 지정된 언어로 번역된 결과 반환

```python
from translate import Translator
```

```python
text = '오늘 실습 어려울 거야. 근데 도움 많이 될거야. 스트롱 스트롱. 상대 세게 나온다. 그런 스트레스도 필요해.'

# Translator 객체 생성
translator = Translator(to_lang='en', from_lang='ko')
```

```python
# 번역 실행
result = translator.translate(text)

print(text)
print(result)
```
