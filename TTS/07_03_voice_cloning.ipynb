# 보이스 클로닝, 내 목소리로 합성하기
- OmniVoice 의 **zero-shot voice cloning**. 
- 짧은 참조 음성(10초 안팎) 과 그 음성에 해당하는 텍스트만 있으면, 새 문장을 그 목소리 톤으로 합성함

### 미리 알아둘 룰
- 참조 음성 길이는 **5~15초** 정도가 적절
- 마이크에 가깝게 (10~20cm), 조용한 환경에서
- 참조 음성에 말한 문장을 **정확히 그대로** `ref_text` 로 넘겨야 함

```python
%pip install Omnivoice
import torch
import soundfile as sf
import numpy as np
from omnivoice import OmniVoice
from IPython.display import Audio
import os
os.makedirs("samples", exist_ok=True)
os.makedirs("outputs", exist_ok=True)

device = "cuda:0" if torch.cuda.is_available() else "cpu"
dtype  = torch.float16 if torch.cuda.is_available() else torch.float32

model = OmniVoice.from_pretrained("k2-fsa/OmniVoice", device_map=device, dtype=dtype)
print("준비 완료. device:", device)
```

## 2. 내 음성 녹음 (sounddevice)

마이크에서 직접 녹음. `sounddevice` 가 없으면 설치. Windows / Mac / Linux 모두 지원.

녹음할 문장 예시 (10초 안에 끝나는 길이로 한 문장 추천):
- "안녕하세요. 제 이름은 ○○○입니다. 인공지능을 공부하고 있습니다. 오늘도 열심히 하겠습니다."

### 녹음이 안 되거나 마이크가 없는 경우

직접 핸드폰·녹음기로 녹음한 wav 파일을 `samples/my_voice.wav` 로 넣고, 위 셀의 `REF_TEXT` 만 그 내용에 맞게 정확히 적어주세요.

## 3. 보이스 클로닝, 새 문장을 내 목소리로

`ref_audio` + `ref_text` + 합성할 새 `text`.

```python
# 먼저 위의 1. 셋업 셀을 실행하여 model 변수를 생성하세요.
REF_TEXT = '안녕하세요. 제 이름은 김도훈입니다. 인공지능 강의를 듣고 있습니다. 오늘도 화이팅입니다.'
new_text = '오늘 우리가 다룰 주제는 OmniVoice를 활용한 보이스 클로닝입니다.'

audio = model.generate(
    text=new_text,
    ref_audio='data/My_voice.wav',
    ref_text=REF_TEXT,
)
sf.write('outputs/clone_01.wav', audio[0], 24000)
print(f"저장: outputs/clone_01.wav  ({len(audio[0])/24000:.1f}초)")
Audio('outputs/clone_01.wav')
```

## 4. 여러 문장을 같은 목소리로

같은 참조 음성으로 다양한 문장을 합성. 내 목소리로 NPC 대사, 인삿말, 광고 멘트 등 만들기.

```python
new_lines = {
    "clone_npc"     : "용사여, 이 검을 받아라. 너의 여정을 응원하겠다.",
    "clone_greeting": "안녕하세요 여러분, 오늘도 좋은 하루 되세요.",
    "clone_promo"   : "지금 가입하면 첫 3개월 무료. 놓치지 마세요.",
    "clone_warning" : "잠깐, 한 번 더 생각해보세요. 정말 이 선택이 맞나요?",
}

for name, text in new_lines.items():
    audio = model.generate(
        text=text,
        ref_audio='./samples/my_voice.wav',
        ref_text=REF_TEXT
        )
    path = f'outputs/{name.wax}'
    sf.write(path, audio[0], 24000)
    print(f"  saved: {path}")
```

```python
for name in new_lines:
    print(name)
    display(Audio(f"outputs/{name}.wav"))
```
