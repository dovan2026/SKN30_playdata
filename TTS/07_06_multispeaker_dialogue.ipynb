# Multi-Speaker Dialogue, 다중 화자 대화

- 게임 NPC 대화·라디오 드라마·인터뷰 같은 **다중 화자 시나리오**를 한 번에 합성.

```python
import re
import torch
import numpy as np
import soundfile as sf
from omnivoice import OmniVoice
from IPython.display import Audio
import os
os.makedirs("outputs", exist_ok=True)

device = "cuda:0" if torch.cuda.is_available() else "cpu"
dtype  = torch.float16 if torch.cuda.is_available() else torch.float32

model = OmniVoice.from_pretrained("k2-fsa/OmniVoice", device_map=device, dtype=dtype)
print("준비 완료. device:", device)
```

## 1. 대화 텍스트 파싱 함수

`[Speaker_N]: 발화 내용` 형식을 (speaker_id, utterance) 리스트로 분해.

```python
SPEAKER_PATTERN = re.compile(r"\[Speaker_(\d+)\]:\s*(.+?)(?=\[Speaker_\d+\]:|$)", re.DOTALL)


def parse_dialogue(text: str):
    """[Speaker_N]: 패턴을 (speaker_id, utterance) 리스트로 분해."""
    turns = []
    for m in SPEAKER_PATTERN.finditer(text):
        speaker_id = int(m.group(1))
        utterance = m.group(2).strip()
        turns.append((speaker_id, utterance))
    return turns

# 테스트
sample = '''
[Speaker_0]: 용사여, 드디어 왔구나. 오랜 시간 기다렸다.
[Speaker_1]: 오래 기다리셨네요. 제가 그 임무를 받으러 왔습니다.
[Speaker_0]: 좋다. 이 검을 받아라. 너의 여정에 함께할 것이다.
[Speaker_1]: 감사합니다. 반드시 임무를 완수하겠습니다.
'''.strip()

for sid, text in parse_dialogue(sample):
    print(f"[Speaker_{sid}] {text[:50]}{'...' if len(text)>50 else ''}")
```

## 2. 화자별 음성 매핑

각 speaker_id 에 음성 모드를 매핑. 두 가지 방식.

### A. Voice Design (instruct) 매핑

녹음 없이 자연어 속성으로 화자 구분.

```python
SPEAKER_DESIGN = {
    0: "male, elderly, low pitch",          # 늙은 마법사
    1: "male, young adult, moderate pitch"  # 젊은 용사
}

def generate_dialogue_design(text: str, speakers: dict, gap_sec: float = 0.4):
    """Voice Design 매핑으로 다중 화자 대화 합성."""
    turns = parse_dialogue(text)
    pieces = []
    for sid, utterance in turns:
        instruct = speakers.get(sid)
        if instruct is None:
            print(f"[skip] Speaker_{sid} 매핑 없음")
            continue
        audio = model.generate(text=utterance, instruct=instruct)
        pieces.append(audio[0])
        # 화자 전환 사이 짧은 정적
        if gap_sec > 0:
            pieces.append(np.zeros(int(gap_sec * 24000), dtype=audio[0].dtype))
    return np.concatenate(pieces, axis=0)

result = generate_dialogue_design(sample, SPEAKER_DESIGN)
sf.write("outputs/dialog_design.wav", result, 24000)
print(f"길이: {len(result)/24000:.1f}초")
Audio("outputs/dialog_design.wav")
```

### B. Voice Cloning (ref_audio) 매핑

화자별로 미리 녹음한 wav 와 ref_text 를 매핑. 사람 목소리를 그대로 재현하고 싶을 때.

```python
# samples/ 폴더에 각 화자별 wav + ref_text 준비 필요
SPEAKER_CLONE = {
    0 : ("/content/my_voice_1.wav", "안녕하세요. 제 이름은 안현준입니다. 인공지능 강의를 맡고 있습니다. 오늘도 화이팅입니다."),
    1 : ("/content/my_voice_2.wav", '안녕하세요 제 이름은 정민규 입니다. 방탄소년단 여덟번째 멤버를 맡고 있습니다.')
}

def generate_dialogue_clone(text: str, speakers: dict, gap_sec: float = 0.4):
    """Voice Cloning 매핑으로 다중 화자 대화 합성."""
    turns = parse_dialogue(text)
    pieces = []
    for sid, utterance in turns:
        mapping = speakers.get(sid)
        if mapping is None:
            print(f"[skip] Speaker_{sid} 매핑 없음")
            continue
        ref_audio, ref_text = mapping
        if not os.path.exists(ref_audio):
            print(f"[skip] {ref_audio} 없음")
            continue
        audio = model.generate(text=utterance, ref_audio=ref_audio, ref_text=ref_text)
        pieces.append(audio[0])
        if gap_sec > 0:
            pieces.append(np.zeros(int(gap_sec * 24000), dtype=audio[0].dtype))
    return np.concatenate(pieces, axis=0) if pieces else None

# samples/my_voice.wav 가 있으면 Speaker_0 만 클로닝, 나머지는 Design fallback
print("Cloning 매핑이 있는 화자만 클로닝, 나머지는 별도 처리 필요")
```

## 3. 게임 시나리오, 마법사 + 용사 + 마을사람

3 화자 대화. 각 화자별 톤을 Voice Design 으로 구분.

```python
script = '''
[Speaker_0]: 자네가 새로 마을에 온 그 용사인가?
[Speaker_1]: 네, 저는 새로 임명된 용사입니다. 마을 분들께 인사 드리러 왔습니다.
[Speaker_2]: 오! 진짜 용사님이세요? 우리 마을 좀 도와주세요. 요즘 늑구가 많아져서 너무 무서워요.
[Speaker_1]: 걱정 마세요. 저는 천하무적이에용.
[Speaker_2]: 잘 부탁하네. 보상은 낡은 지팡이 하나 주겠네. 이따 마을 회관으로 오시오.
'''.strip()

SPEAKERS_3 = {
    0: 'male, elderly, low pitch',              # 마을 촌장(늙은 남성)
    1: 'male, young adult, moderate pitch',     # 용사 (젊은 남성)
    2: 'female, young adult, high pitch'        # 마을 소녀
}

result = generate_dialogue_design(script, SPEAKERS_3)
sf.write("outputs/dialog_village.wav", result, 24000)
print(f"길이: {len(result)/24000:.1f}초")
Audio("outputs/dialog_village.wav")
```

## 4. 비언어 심볼과 결합

각 대사에 비언어 심볼 (`[laughter]` 등) 도 넣을 수 있습니다. 단 `[Speaker_N]:` 와 헷갈리지 않게 발화 부분에만.

```python
emotional_script = '''
[Speaker_0]: [laughter] 또 니가? 장난 똥이가! 저기로 가줄래?.
[Speaker_1]: [sigh] 꼭 그래야만 속이 후련했냐! 이 해바라기야.
[Speaker_0]: [dissatisfaction-hnn] 한번 붙자는 거지? 들어와주세요.
'''.strip()

SPEAKERS_4 = {
    0: 'male, young adult, high pitch',
    1: 'male, young adult, high pitch'
}

result = generate_dialogue_design(emotional_script, SPEAKERS_4)
sf.write("outputs/dialog_emotional.wav", result, 24000)
Audio("outputs/dialog_emotional.wav")
```

## 6. 정리

- `parse_dialogue(text)` 로 `[Speaker_N]:` 텍스트를 리스트로 분해
- 화자별 `instruct` (Voice Design) 또는 `ref_audio` (Voice Cloning) 매핑
- 화자 전환 사이 짧은 정적 (`np.zeros`) 으로 자연스러움
- `np.concatenate` 로 모든 발화 이어붙임

## [실습]
1. 화자 2명짜리 짧은 대화 (4~6턴) 를 직접 적어 합성.
2. 화자 전환 사이 정적 길이 (`gap_sec`) 를 0.2 / 0.6 / 1.0 으로 바꿔 자연스러움 비교.
3. 화자 4명짜리 회의 장면을 만들어 각자 다른 `instruct` 로 합성.
4. Voice Cloning 매핑으로 본인 + 친구 (각각 ref_audio 녹음) 두 사람의 가짜 대화 만들기.

```python
shot_script = '''
[Speaker_0]: [question-ah] 아니, 벌써 퇴근 시간이라고? 믿을 수가 없네.
[Speaker_1]: [laughter] 어휴, 과장님. 집중 좀 하세요! 벌써 시계는 5시를 넘었는걸요.
[Speaker_0]: [sigh] 하아, 오늘도 일이 산더미인데... 언제 다 한담.
[Speaker_1]: [confirmation-en] 괜찮아요, 제가 조금 도와드릴게요! 같이 하면 금방 끝날 거예요.
[Speaker_0]: [surprise-oh] 오! 정말? 자네는 정말 천사 같군!
[Speaker_1]: [dissatisfaction-hnn] 헤헷, 그럼 이제 빨리 마무리하시죠. 저도 칼퇴해야 하니까요.
'''.strip()

SPEAKERS_5 = {
    0: 'male, young adult, high pitch',
    1: 'female, young adult, high pitch'
}

result = generate_dialogue_design(shot_script, SPEAKERS_5, gap_sec=0.6)
sf.write("outputs/dialog_shot.wav", result, 24000)
Audio("outputs/dialog_shot.wav")
```
