```python
from langchain_community.document_loaders import PyPDFLoader

# PDF 파일 처리
pdf_path = 'data2/스마트팜 질의응답.pdf'    # pdf 파일 경로 입력
loader = PyPDFLoader(pdf_path)            # pdf 파일 읽기
documents = loader.load()

documents
```

```python
from langchain_community.document_loaders import WebBaseLoader

website_loader = WebBaseLoader("https://www.rda.go.kr/webzine/2022/12/sub1-7.html")
website_documents = website_loader.load()
print(website_documents)
```

```python
# 흩어진 Document 합치기
full_text = ""

for doc in documents:
    full_text += doc.page_content + "\n"

print(full_text)
```

```python
import re
from langchain_core.documents import Document

pattern = r"Q:.*?\nA:.*?(?=\nQ:|\Z)"  # Q: ~ A: ~ 다음 Q: 전까지

matches = re.findall(pattern, full_text, re.DOTALL)

# Document 객체로 변환
qa_documents = []

for chunk in matches:    
    clean_text = chunk.strip()  # 앞뒤 공백 제거        
    doc = Document(page_content=clean_text)  # LangChain용 문서 객체로 감싸기
    qa_documents.append(doc)  # 리스트에 추가

qa_documents
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 텍스트 분할
text_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", "Q:", "A:"],  # 단락, 줄바꿈, Q/A 기준 모두 활용
    chunk_size=200,       # Q&A 1~2쌍이 들어갈 정도
    chunk_overlap=50,     # 중요한 문장이 끊기지 않도록 약간의 겹침 유지
)

texts = text_splitter.split_documents(website_documents)
print(f"총 {len(texts)}개의 텍스트 청크로 분할되었습니다.")
```

```python
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS



# 벡터 스토어 생성
# 텍스트를 OpenAI용 벡터로 변환
embeddings = OpenAIEmbeddings()

# FAISS가 벡터를 받아 인덱스를 만들고 빠르게 검색할 수 있도록 준비
vectorstore= FAISS.from_documents(texts, embeddings)
```

```python
vectorstore.index.ntotal
```

```python
# 검색 설정
retriever = vectorstore.as_retriever(search_kwargs={'k':2}) # 유사도가 가장 높은 문서 2개만 가져오도록
```

```python
data = retriever.invoke('스마트 팜 현장지원센터란?')    # 질문을 벡터로 임베딩 -> FAISS 검색 -> 가장 가까운 2개의 문서 변환
data
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough # 질문을 텍스트 그대로 넘기기

def format_docs(docs):
    return '\n\n'.join(doc.page_content for doc in docs)

# 프롬프트 생성
prompt = PromptTemplate.from_template(
    '''당신은 친절한 질의응답 도우미입니다.
    다음 검색된 맥락을 사용하여 질문에 답하세요.
    답을 모르면 모른다고 말하세요.
    한국어로 답하세요.
    
    #Question: {question}
    #Context: {context}
    #Answer:'''
)

# 모델(LLM)을 생성합니다.
llm = ChatOpenAI(model='gpt-4.1-mini', temperature=0)

# 체인(chain) 생성
chain = (
    {
        'context':retriever | format_docs, 'question':RunnablePassthrough()
    }
    | prompt
    | llm
    | StrOutputParser()
    )
```

```python
result = chain.invoke('대표적인 우리나라 스마트농업 기술')

print(result)
```
