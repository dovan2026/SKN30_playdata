
# 아이템 기반 협업 필터링

- **아이템 간의 유사도를 분석** 하여 **사용자가 좋아할 만한 아이템을 예측** 하는 기법

```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = {
    "사용자": ['A', "B", "C", "D", "E"],
    "기생충": [5,4,1,0,3],
    "부산행": [0,2,0,4,0],
    "태극기 휘날리며": [3, 0, 5, 4, 0],
    "도둑들" : [1, 0, 4, 2, 4],
    "설국열차" : [0, 3, 0, 0 ,5],
    "범죄도시" : [2, 0, 4, 0, 3]
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df = df.set_index('사용자')
```

```python
df.head()
```

```python
# 코사인 유사도는 행간의 유사도를 구할 수 있기 때문에 행, 열을 바꿔줌
item_user_matrix = df.T
item_user_matrix.head()
```

```python
# 코사인 유사도 계산
item_sim = cosine_similarity(item_user_matrix)
item_sim_df = pd.DataFrame(item_sim,
                           index=df.columns,
                           columns=df.columns )
item_sim_df
```

```python
# 특정 사용자(A)
user = 'A'
user_rating = df.loc[user]
user_rating
```

```python
# 점수가 0인. 아직 보지 않은 아이템
user_unrated_items = user_rating[user_rating==0].index.tolist()     # df에서 'A'유저의 행(평점 데이터)만 가져옴
user_unrated_items
```

```python
recommend_dic = {}      # 추천 영화 및 예상 평점을 담을 딕셔너리

for item in user_unrated_items: # A유저 : 부산행, 설국열차
    sim_score = item_sim_df[item]       # 현재 아이템에 대한 유사도 점수 가져오기

    user_rated_items = user_rating[user_rating > 0].index.tolist()  # 사용자가 평점을 부여한 목록

    sim_score = sim_score[user_rated_items]     # 사용자가 평점을 부여한 아이템에 대한 유사도 점수를 필터링

    user_rating_filered = user_rating[user_rated_items]     # 사용자가 평점을 부여한 아이템의 평점만 가져오기

    weight_sum = sum(sim_score * user_rating_filered)       # 유사도 * 사용자 평점

    sim_sum = sum(sim_score)            # 유사도 총합
    print(weight_sum, sim_sum)          # weight_sum: 현재 추천 후보인 영화가 기존에 유저가 평가한 영화들과 얼마나 관련 있는지에 대한 가중 점수
                                        # sim_sum : 추천하려는 영화가 기존에 본 영화들과 더 유사하다는 의미

    if sim_sum != 0:
        recommend_dic[item] = weight_sum/sim_sum
    else:
        recommend_dic[item] = 0         # 유사도가 없으면 추천 점수 0
```
