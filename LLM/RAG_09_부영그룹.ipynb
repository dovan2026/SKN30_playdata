# 1. LOAD

```python
from langchain_community.document_loaders import PyPDFLoader

# PDF 파일 처리
pdf_path = "data2/부영그룹.pdf"  # PDF 파일 경로 입력
loader = PyPDFLoader(pdf_path)  # PDF 파일 읽기
documents = loader.load()
print(documents)
```

```python
from langchain_community.document_loaders import WebBaseLoader

website_loader = WebBaseLoader("https://n.news.naver.com/article/437/0000378416")
website_documents = website_loader.load()
print(website_documents)
```

# 2. SPLIT

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 텍스트 분할
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,   # 청크의 최대 크기, 예)GPT-4의 입력 토큰 한계(8100~32000 토큰). 문자 수 기준
    chunk_overlap=50  # 청크 사이의 중복 정도. 청큰 간 겹침(문맥 보존)
)

texts = text_splitter.split_documents(documents)
print(f"총 {len(texts)}개의 텍스트 청크로 분할되었습니다.")
```

# 3. EMBED, STORE

```python
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS
from dotenv import load_dotenv
import os

load_dotenv()

# 벡터 스토어 생성
# 텍스트를 OpenAI용 벡터로 변환합니다.
embeddings = OpenAIEmbeddings()
# FAISS가 벡터를 받아 인덱스를 만들고 빠르게 검색할 수 있도록 준비
db = FAISS.from_documents(texts, embeddings)
```

```python
# 질문
question = "부영그룹에서 제공하는 직원 혜택은?"
```
